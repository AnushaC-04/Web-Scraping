### to get all the details do this 

```python
import requests
from bs4 import BeautifulSoup

names,prices,ratings,availability=[],[],[],[]
```

```python
for i in range(1,51):
    url=f'https://books.toscrape.com/Catalogue/page-{i}.html'
    res=requests.get('https://books.toscrape.com/')
    code=res.text
    soup=BeautifulSoup(code)


    arti_tag=soup.find_all('article')

    for article in arti_tag:
        names.append(article.find('h3').find('a').attrs['title'] )
        prices.append(article.find('div',attrs={'class':'product_price'}).find('p').text)
        ratings.append(article.find('p').attrs['class'][-1])
        availability.append(article.find('div',attrs={'class':'product_price'}).find('p',attrs={'class':'instock availability'}).text)
```

```python
import pandas as pd
d={'title':names,'price':prices,'rating':ratings,'availability':availability}
df=pd.DataFrame(d)
df.to_excel('books_1000.xlsx',index=False)
```
