```python
#  driver--> whole webpage
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time

driver=webdriver.Chrome()
time.sleep(2)

driver.get('https://www.flipkart.com/')
time.sleep(2)

driver.find_element(By.CLASS_NAME,'b3wTlE').click()   
time.sleep(2)

search_box=driver.find_element(By.NAME,'q')
search_box.send_keys('samsung mobiles')
search_box.send_keys(Keys.ENTER)
time.sleep(3)

img_xpath='//*[@id="container"]/div/div[3]/div[1]/div[2]/div[2]/div/div/div/a/div[2]/div[1]/div/div/img'
img_tag=driver.find_element(By.XPATH,img_xpath)
img_tag.screenshot('samsung_images/s_gal_f70.jpg')

mob_names,mob_prices=[],[]
for i in range(3):

    time.sleep(3)  # without time if net is slow it will show eroor because next buttuon is not loaded
    div_lst=driver.find_elements(By.CLASS_NAME,'jIjQ8S')
       
    for div in div_lst:
        try:
            mob_names.append(div.find_element(By.CLASS_NAME,'RG5Slk').text)
        except:
            mob_names.append(None)
        try:
            mob_prices.append(div.find_element(By.CSS_SELECTOR,'.hZ3P6w.DeU9vF').text) 
        except:
            mob_prices.append(None)

    nxt_link=driver.find_elements(By.CLASS_NAME,'jgg0SZ')[-1].click()  

time.sleep(2)
driver.quit()
```
